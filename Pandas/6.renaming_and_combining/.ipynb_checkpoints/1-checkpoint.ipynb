```python
import pandas as pd
reviews = pd.read_excel("./reviews.xlsx")
pd.set_option('display.max_rows', 5)
```

```python
# reviews: Refers to the Pandas DataFrame containing your data.
# .rename(columns={'points': 'score'}): Renames the column 'points' to 'score' within the DataFrame.

#     columns specifies that you're renaming columns.
#     {'points': 'score'} is a dictionary mapping the old column name to the new one.

reviews.rename(columns={'points': 'score'})
```

```python
# reviews: Refers to the Pandas DataFrame containing your data.
# .rename(index={0: 'firstEntry', 1: 'secondEntry'}): Renames the index labels 0 and 1 to 'firstEntry' and 'secondEntry' respectively.

#     index specifies that you're renaming index labels.
#     0: 'firstEntry' and 1: 'secondEntry' map the old index labels to the new ones.
    
reviews.rename(index={0: 'firstEntry', 1: 'secondEntry'})
```

```python
1. side by side
    1. axis=1
    2. merge or join
2. stacking
    1. axis=0
    2. concat or append
```

```python

axis=1 in pd.concat() means that you're concatenating DataFrames along the columns.
Breakdown:

    axis=0 (default): Concatenates DataFrames vertically, stacking them one on top of the other.
    axis=1: Concatenates DataFrames horizontally, placing them side by side.
```

```python


Visual Representation:

axis=0 (default):

df1:
   A  B
0  1  3
1  2  4

df2:
   A  B
0  5  7
1  6  8

Concatenated DataFrame:
   A  B  A  B
0  1  3  5  7
1  2  4  6  8
```

```python
axis=1:

df1:
   A  B
0  1  3
1  2  4

df2:
   A  B
0  5  7
1  6  8

Concatenated DataFrame:
   A  B  A  B
0  1  3  5  7
1  2  4  6  8
```

```python
import pandas as pd

df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
print(df1)
print(df2)

# Concatenate along rows
df_concat_rows = pd.concat([df1, df2], axis=0)
print('======concat=====axis=0, ==')
print(df_concat_rows)

# Concatenate along columns
df_concat_cols = pd.concat([df1, df2], axis=1)
print('======concat=====axis=1, ==')
print(df_concat_cols)
```

```python
import pandas as pd

# Sample DataFrames
df1 = pd.DataFrame({'key': ['K0', 'K1', 'K2'], 'A': ['A0', 'A1', 'A2'], 'B': ['B0', 'B1', 'B2']})
df2 = pd.DataFrame({'key': ['K1', 'K2', 'K3'], 'C': ['C0', 'C1', 'C2'], 'D': ['D0', 'D1', 'D2']})
print(df1)
print(df2)

merged_inner = pd.merge(df1, df2, on='key', how='inner')
print('============merged_inner===============')
print(merged_inner)

merged_left = pd.merge(df1, df2, on='key', how='left')
print('============merged_left===============')
print(merged_left)

merged_right = pd.merge(df1, df2, on='key', how='right')
print('============merged_right===============')
print(merged_right)
```

```python
merged_left = pd.merge(df1, df2, left_on='key', right_on='key', how='left')
print('============merged_left===============')
print(merged_left)
```
