```python
Key Features of bytes:

    Immutable – Once created, it cannot be modified.
    Binary Representation – Prefixed with b to indicate bytes.
    Supports Indexing and Slicing – Like strings, you can access individual bytes.
    Cannot Contain Non-ASCII Characters – Only ASCII characters (0-255) are allowed.
```

# List

| Feature                  | Tuple                                       | List                                        |
|--------------------------|---------------------------------------------|---------------------------------------------|
| **Definition**            | Ordered, immutable collection of items      | Ordered, mutable collection of items       |
| **Syntax**                | Parentheses ( )                             | Square brackets [ ]                        |
| **Mutability**            | Immutable (cannot be changed after creation) | Mutable (can be changed after creation)    |
| **Methods**               | Limited methods (e.g., count(), index())    | Rich set of methods (e.g., append(), remove(), pop(), sort()) |
| **Use Cases**             | Suitable for read-only data                 | Suitable for data that may need to be modified |
| **Performance**           | Generally faster due to immutability        | May be slower due to mutability            |
| **Memory**                | May use slightly less memory due to immutability | May use slightly more memory               |
| **Immutability Benefits** | Safer to use as dictionary keys and in sets | Not safe for use as dictionary keys or in sets |


```python
# Types of Built-in Data Structures in Python: List, Tuple, Sets, Dictionary.
# https://www.edureka.co/blog/data-structures-in-python/#list
# Creating a list
my_list = [] #create empty list
print(my_list)
my_list = [1, 2, 3, 'example', 3.132] #creating list with data
print(my_list)
```

```python
# Creating a list
my_list = [1, 2, 3]
print(my_list)  # Output: [1, 2, 3]

# The append() function adds all the elements passed to it as a single element.
my_list.append([555, 12])  # Adds [555, 12] as a single list inside my_list
print(my_list)  # Output: [1, 2, 3, [555, 12]]

# The extend() function adds the elements one-by-one into the list.
my_list.extend([234, 'more_example'])  # Adds 234 and 'more_example' separately
print(my_list)  # Output: [1, 2, 3, [555, 12], 234, 'more_example']

# The insert() function adds an element at the specified index.
my_list.insert(1, 'insert_example')  # Inserts 'insert_example' at index 1
print(my_list)  # Output: [1, 'insert_example', 2, 3, [555, 12], 234, 'more_example']
```

```python
# Deleting Elements
my_list = [1, 2, 3, 'example', 3.132, 10, 30]

# To delete elements, use the del keyword which is built-in into Python but this does not return anything back to us.
del my_list[5] #delete element at index 5 [starts from 0 index]
print(my_list)  # Output: [1, 2, 3, 'example', 3.132, 30]

# To remove an element by its value, you use the remove() function.
my_list.remove('example') #remove element with value
print(my_list)  # Output: [1, 2, 3, 3.132, 30]

# If you want the element back, you use the pop() function which takes the index value.
a = my_list.pop(1) #pop element from list
print('Popped Element: ', a, ' List remaining: ', my_list)  # Output: Popped Element: 2  List remaining: [1, 3, 3.132, 30]

# Empty the list
my_list.clear() #empty the list
print(my_list)  # Output: []
```

```python
# Accessing Elements
my_list = [1, 2, 3, 'example', 3.132, 10, 30]

# Access elements one by one
for element in my_list:
    print(element)
# Output:
# 1
# 2
# 3
# example
# 3.132
# 10
# 30

# Access all elements
print(my_list) 
# Output: [1, 2, 3, 'example', 3.132, 10, 30]

# Access element at index 3
print(my_list[3])
# Output: example

# Access elements from index 0 to 1 (excluding 2)
print(my_list[0:2]) 
# Output: [1, 2]

# Access elements in reverse order
print(my_list[::-1])
# Output: [30, 10, 3.132, 'example', 3, 2, 1]
```

```python
# Other Functions
my_list = [1, 2, 3, 10, 30, 10]

# Find the length of the list
print(len(my_list)) 
# Output: 6

# Find the index of the first occurrence of the element '30'
print(my_list.index(30)) 
# Output: 4

# Count the number of occurrences of the element '10'
print(my_list.count(10)) 
# Output: 2

# Print the sorted list without changing the original list
print(sorted(my_list)) 
# Output: [1, 2, 3, 10, 10, 30]

# Sort the original list in descending order
my_list.sort(reverse=True) 
print(my_list) 
# Output: [30, 10, 10, 3, 2, 1]
```

# Loop

```python
# Iterating over a list
my_list = [1, 2, 3, 4, 5]
for element in my_list:
    print(element)
# Output: 1 2 3 4 5

# Iterating over a string
my_string = "hello"
for char in my_string:
    print(char)
# Output: h e l l o
```

```python
# Using a while loop to print numbers from 1 to 5
counter = 1
while counter <= 5:
    print(counter)
    counter += 1
# Output: 1 2 3 4 5
```

```python
# Using break in a for loop
for num in range(10):
    if num == 5:
        break
    print(num)
# Output: 0 1 2 3 4

# Using continue in a for loop
for num in range(10):
    if num % 2 == 0:
        continue
    print(num)
# Output: 1 3 5 7 9
```

```python
# Nested for loop
for i in range(3):
    for j in range(3):
        print(f"i={i}, j={j}")

# F-String: The f before the string indicates that it is an f-string.
# Curly Braces {}: The expressions inside the curly braces {} are evaluated at runtime and their values are inserted into the string.

# Output: 
# i=0, j=0
# i=0, j=1
# i=0, j=2
# i=1, j=0
# i=1, j=1
# i=1, j=2
# i=2, j=0
# i=2, j=1
# i=2, j=2
```

# Tuple

```python
# Creating a Tuple
my_tuple = (1, 2, 3)  # A tuple with three elements
print(my_tuple)
# Output: (1, 2, 3)

# Adding elements to the tuple by concatenation
my_tuple = my_tuple + (4, 5, 6)  # Concatenating tuples to add new elements
print(my_tuple)
# Output: (1, 2, 3, 4, 5, 6)
```

```python
# Accessing Elements in Tuple
my_tuple = (1, 2, 3, 'Dewan Farid')  # Define a tuple with different types of elements

# Access and print elements one by one
for x in my_tuple:
    print(x)
# Output:
# 1
# 2
# 3
# Dewan Farid

# Print the entire tuple
print(my_tuple)
# Output: (1, 2, 3, 'Dewan Farid')

# Access and print the element at index 1
print(my_tuple[1])
# Output: 2

# Access and print all elements (complete tuple)
print(my_tuple[:])
# Output: (1, 2, 3, 'Dewan Farid')

# Access and print the 5th character of the element at index 3 (which is 'Dewan Farid')
print(my_tuple[3][4])
# Output: n
```

```python
# Other Functions
my_tuple = (1, 2, 3, ['java', 'python'])
print(my_tuple) 
# Output: (1, 2, 3, ['java', 'python'])

# Modify an element inside the mutable list within the tuple
my_tuple[3][0] = 'julia'
print(my_tuple) 
# Output: (1, 2, 3, ['julia', 'python'])

# Count the number of times '2' appears in the tuple
print(my_tuple.count(2)) 
# Output: 1

# Find the index of the sublist ['julia', 'python']
print(my_tuple.index(['julia', 'python']))
# Output: 3
```

# SET

```python
# Creating a set
my_set = {1, 2, 3, 4, 5, 5, 5}
print(my_set)
# Output: {1, 2, 3, 4, 5}

# Adding elements to the set
my_set.add(7)
print(my_set)
# Output: {1, 2, 3, 4, 5, 7}
```

```python
# Operations in sets
# The union() function combines the data present in both sets.
# The intersection() function finds the data present in both sets only.
# The difference() function deletes the data present in both and outputs data present only in the set passed.
# The symmetric_difference() does the same as the difference() function but outputs the data which is remaining in both sets.

my_set = {1, 2, 3, 4}
my_set_2 = {3, 4, 5, 6}

# Union: Combines elements of both sets
print(my_set.union(my_set_2), '----------', my_set | my_set_2)
# Output: {1, 2, 3, 4, 5, 6} ---------- {1, 2, 3, 4, 5, 6}

# Intersection: Elements present in both sets
print(my_set.intersection(my_set_2), '----------', my_set & my_set_2)
# Output: {3, 4} ---------- {3, 4}

# Difference: Elements in my_set but not in my_set_2
print(my_set.difference(my_set_2), '----------', my_set - my_set_2)
# Output: {1, 2} ---------- {1, 2}

# Symmetric Difference: Elements in either set, but not in both
print(my_set.symmetric_difference(my_set_2), '----------', my_set ^ my_set_2)
# Output: {1, 2, 5, 6} ---------- {1, 2, 5, 6}

# Clear: Empties the set
my_set.clear()
print(my_set)
# Output: set()
```

# Dictionaries

```python
# Dictionaries are used to store key-value pairs.
my_dict = {}  # Create an empty dictionary
print(my_dict)
# Output: {}

# Create a dictionary with elements
my_dict = {1: 'Python', 2: 'Java'}  # Dictionary with keys 1 and 2, and values 'Python' and 'Java'
print(my_dict)
# Output: {1: 'Python', 2: 'Java'}
```

```python
# Changing and Adding key, value pairs
my_dict = {'First': 'Python', 'Second': 'Java'}  # Initialize dictionary with two key-value pairs
print(my_dict)
# Output: {'First': 'Python', 'Second': 'Java'}

# Change the value of the key 'Second'
my_dict['Second'] = 'C++'  # Change the value of the key 'Second' to 'C++'
print(my_dict)
# Output: {'First': 'Python', 'Second': 'C++'}

# Add a new key-value pair
my_dict['Third'] = 'Julia'  # Add a new key-value pair 'Third': 'Julia'
print(my_dict)
# Output: {'First': 'Python', 'Second': 'C++', 'Third': 'Julia'}

# Accessing Elements
print(my_dict['First'])  # Access the value associated with the key 'First'
# Output: Python

print(my_dict.get('Second'))  # Access the value associated with the key 'Second' using the get() method
# Output: C++

# Other Functions
print(my_dict.keys())  # Get all the keys in the dictionary
# Output: dict_keys(['First', 'Second', 'Third'])

print(my_dict.values())  # Get all the values in the dictionary
# Output: dict_values(['Python', 'C++', 'Julia'])

print(my_dict.items())  # Get all the key-value pairs in the dictionary
# Output: dict_items([('First', 'Python'), ('Second', 'C++'), ('Third', 'Julia')])

print(my_dict.get('First'))  # Access the value associated with the key 'First' again using the get() method
# Output: Python
```

```python
# Deleting key, value pairs
# To delete the values, you use the pop() function which returns the value that has been deleted.
# To retrieve the key-value pair, you use the popitem() function which returns a tuple of the key and value.
# To clear the entire dictionary, you use the clear() function.

my_dict = {'First': 'Python', 'Second': 'Java', 'Third': 'Julia'}

# Pop element with key 'Third'
a = my_dict.pop('Third')
print('Value:', a)
print('Dictionary:', my_dict)
# Output:
# Value: Julia
# Dictionary: {'First': 'Python', 'Second': 'Java'}

# Pop the last key-value pair
b = my_dict.popitem()
print('Key, value pair:', b)
print('Dictionary', my_dict)
# Output:
# Key, value pair: ('Second', 'Java')
# Dictionary: {'First': 'Python'}

# Clear the entire dictionary
my_dict.clear()
print('Dictionary', my_dict)
# Output:
# Dictionary: {}
```
