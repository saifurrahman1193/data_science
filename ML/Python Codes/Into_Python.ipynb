```python
var = int(input("Enter an Integer: "))
if var % 2 == 0: 
  print("Even")
else: 
  print("Odd")
```

```python
# Function declaration
def evenOdd():
  var = int(input("Enter an Integer: "))
  if var % 2 == 0: 
    print("Even")
  else: 
    print("Odd")
  return 
```

```python
evenOdd()
```

```python
# Creating a Class in Python:
class SuperHero:
  def power(self):
    print("Superhero has some power")
    return
```
