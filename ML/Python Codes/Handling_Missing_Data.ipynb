```python
import pandas as pd
import numpy as np
import io
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
diabetes_df = pd.read_csv(io.BytesIO(uploaded["diabetes.csv"]))
display(diabetes_df)
```

```python
data = diabetes_df.drop(['Outcome'], axis = 1)
```

```python
# Replacing 0 with nan values
data.replace(0, np.nan, inplace = True)
```

```python
data.isnull().sum().sort_values(ascending = False)
```

```python
# Handling missing data
data ['Insulin'].unique()
```

```python
np.round(data['Insulin'].mean())
```

```python
data ['Insulin'].replace(np.nan, np.round(data['Insulin'].mean()), inplace = True)
```

```python
from sklearn.impute import SimpleImputer
# strategy = 'median', 'mean'
impute = SimpleImputer(strategy = 'median')
data_array = impute.fit_transform(data)
```
