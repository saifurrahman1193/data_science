```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold, cross_val_score
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import make_scorer, mean_squared_error

# Sample data (same as your LaTeX table)
data = {
    'date': ['2024-09-24', '2024-09-25', '2024-09-26', '2024-09-27', '2024-09-28', '2024-09-29',
             '2024-09-30', '2024-10-01', '2024-10-02', '2024-10-03', '2024-10-04', '2024-10-05',
             '2024-10-06', '2024-10-07', '2024-10-08', '2024-10-09', '2024-10-10', '2024-10-11',
             '2024-10-12', '2024-10-13'],
    'balance': [138706, 138214, 137241, 134192, 118229, 116883, 113839, 112638, 111951, 110807,
                105434, 103655, 100337, 96591, 86970, 162602, 155813, 155048, 154230, 154108]
}

df = pd.DataFrame(data)

# Convert date to ordinal for regression purposes
df['date'] = pd.to_datetime(df['date']).map(pd.Timestamp.toordinal)

# Define X (date) and y (balance)
X = df[['date']]
y = df['balance']

# Polynomial transformation (degree 2)
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

# Create the model
model = LinearRegression()

# Define k-fold cross-validation with k=5
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Custom scorer for MSE
mse_scorer = make_scorer(mean_squared_error)

# Perform cross-validation and calculate MSE for each fold
mse_scores = cross_val_score(model, X_poly, y, cv=kf, scoring=mse_scorer)

# Display the MSE for each fold
for i, mse in enumerate(mse_scores, 1):
    print(f'Fold {i} - MSE: {mse}')

# Calculate and display the average MSE across all folds
average_mse = np.mean(mse_scores)
print(f'Average MSE: {average_mse}')
```
