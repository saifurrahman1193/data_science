```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
import string
import re
from nltk import pos_tag, ne_chunk
```

```python
# Download necessary NLTK datasets
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('maxent_ne_chunker')
```

```python
# Sample text
text = " http://saifur.com Naturall Naturall Naturall $ Language Processing (NLP) is a fascinating field. It involves understanding and generating human language! <b>bold</b>"
text
```

```python
# 1. Convert text to lowercase
text = text.lower()
text
```

```python
# remove leading and trailing white space
text = text.strip()

# replace multiple consecutive white space characters with a single space
text = " ".join(text.split())
text
```

```python
# Remove URLs
# define a regular expression pattern to match URLs
pattern = r"(http|ftp|https)://([\w_-]+(?:(?:\.[\w_-]+)+))([\w.,@?^=%&:/~+#-]*[\w@?^=%&/~+#-])?"

# replace URLs with an empty string
text = re.sub(pattern, "", text)
text
```

```python
# Remove HTML code
# define a regular expression pattern to match HTML tags
pattern = r"<[^>]+>"

# replace HTML tags with an empty string
text = re.sub(pattern, "", text)
text
```

```python
# 2. Remove punctuation, special characters
text = text.translate(str.maketrans('', '', string.punctuation))
text
```

```python
# 3. Tokenize text
tokens = word_tokenize(text)
tokens
```

```python
# Remove frequent words. Not necessary for alltime
# calculate the frequency of each word
# fdist = nltk.FreqDist(tokens)

# # remove the most common words (e.g., the top 10% of words by frequency)
# tokens = [token for token in tokens if fdist[token] < fdist.N() * 0.1]
# tokens
```

```python
# 4. Remove stop words
stop_words = set(stopwords.words('english'))
tokens = [word for word in tokens if word not in stop_words]
# filtered_tokens = []
# for word in tokens:
#     if word not in stop_words:
#         filtered_tokens.append(word)
tokens
```

```python
# # Spelling correction
# # get list of English words
# words = nltk.corpus.words.words()

# # correct spelling of each word
# corrected_tokens = []
# for token in tokens:
#     # find the word with the lowest edit distance
#     corrected_token = min(words, key=lambda x: nltk.edit_distance(x, token))
#     corrected_tokens.append(corrected_token)

# token=corrected_tokens
# token
```

```python
# create stemmer object
stemmer = nltk.stem.PorterStemmer()

# stem each token
tokens = [stemmer.stem(token) for token in tokens]
tokens
```

```python
# 5. Lemmatize tokens. also spelling checked here
lemmatizer = WordNetLemmatizer()
tokens = [lemmatizer.lemmatize(word) for word in tokens]
# lemmatized_tokens = []  # Create an empty list to store results
# for word in filtered_tokens:  # Loop through each word in filtered_tokens
#     lemma = lemmatizer.lemmatize(word)  # Apply lemmatization to the word
#     lemmatized_tokens.append(lemma)  # Add the lemmatized word to the list
tokens
```

```python
# Example to Remove Consecutive Duplicate Words
# Function to remove consecutive duplicates
def remove_consecutive_duplicates(tokens):
    result = []
    for i in range(len(tokens)):
        # If the current word is not the same as the previous one, add it to the result
        if i == 0 or tokens[i] != tokens[i - 1]:
            result.append(tokens[i])
    return result

# Remove consecutive duplicates
tokens = remove_consecutive_duplicates(tokens)
tokens
```

```python

# Final processed text
processed_text = ' '.join(tokens)

print("Original Text:")
print(text)
print("\nProcessed Text:")
print(processed_text)
```
