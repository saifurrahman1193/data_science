<h3>Create bag of words representation using CountVectorizer</h3>

```python
from sklearn.feature_extraction.text import CountVectorizer

v = CountVectorizer()
```

Explanation:

    Imports CountVectorizer, which converts text into a bag-of-words matrix.

    v = CountVectorizer() creates the vectorizer object that will be used to extract features from the text data.

```python
emails = [
            'Upto 20% discount reward! Upto',
            'Hey Aksadur, discount Upto'
         ]

emails_cv = v.fit_transform(emails)
emails_cv
```

Explanation:

    Creates a list emails with two example email strings.

    Applies CountVectorizer (v) to fit and transform this list into a sparse matrix emails_cv, which contains word counts (bag-of-words).

    emails_cv is a sparse matrix representation of word frequencies for each email.

```python
emails_cv.toarray()
```

This is a Bag of Words (BoW) matrix generated by CountVectorizer, where:

    Each row = an email.

<h3>Train the naive bayes model</h3>

```python
from sklearn.naive_bayes import MultinomialNB
model = MultinomialNB()
model.fit(X_train_cv, y_train)
```

Explanation
📌 from sklearn.naive_bayes import MultinomialNB

    Imports the Multinomial Naive Bayes classifier from Scikit-learn.

    This algorithm is ideal for text classification problems where features are discrete (like word counts or frequencies).

📌 model = MultinomialNB()

    Initializes a Naive Bayes model instance.

    This model will learn to differentiate between classes (spam vs ham) based on word frequencies.

📌 model.fit(X_train_cv, y_train)

    Trains the model on the training data:

        X_train_cv = bag-of-words representation of training messages.

        y_train = labels (e.g., spam or ham).

    After this step, the model has learned the probability distributions of words given each class.

<h3>Evaluate Performance</h3>

```python
X_test_cv = v.transform(X_test)
```

Explanation

    X_test contains the test email messages (raw text).

    v is the fitted CountVectorizer (trained on X_train), which already knows the vocabulary.

🔧 What this line does:

    Transforms the test messages into the same Bag-of-Words format as the training data (X_train_cv).

    It does not call .fit() again, which is important because:

        .fit() would learn a new vocabulary, possibly different from the training set.

        .transform() uses the same word index mapping learned during training, ensuring consistency.

```python
X_test_cv.toarray()
```

Explanation

    X_test_cv is a sparse matrix (efficient memory-saving structure) representing the Bag-of-Words vector of the test emails.

    Calling .toarray() converts it into a dense NumPy array, which:

        Makes it human-readable

        Shows the actual word counts for each test message

        Uses the same vocabulary and column order as the training data (X_train_cv)

📊 Example Output (Hypothetical)
| Word1 | Word2 | ... | "discount" | "reward" | "upto" |
| ----- | ----- | --- | ---------- | -------- | ------ |
| 0     | 1     | ... | 1          | 0        | 2      |
| 1     | 0     | ... | 0          | 1        | 0      |


Each row = one test email
Each column = one word in the vocabulary
Each value = count of that word in the message

```python
X_test_cv.toarray().shape
```

🧠 Explanation

This returns the shape (dimensions) of the dense array created from your test data's bag-of-words representation.
📊 What the Output Means

The result will be something like:

(n_samples, n_features)

Where:

    n_samples = number of test emails (e.g., 1113 if you did 80/20 train-test split on ~5572 total rows)

    n_features = number of unique words (vocabulary size) learned during vectorization on the training data

✅ Example

(1115, 7456)

    1115 test messages

    7456 unique words (features) learned from training data

```python
model.score(X_test_cv, y_test)
```

Explanation

    This line evaluates the accuracy of your trained MultinomialNB model on the test dataset.

    Specifically, it returns the mean accuracy:
    Accuracy=Number of correct predictionsTotal number of predictions
    Accuracy=Total number of predictionsNumber of correct predictions​

📊 Internally Equivalent To

from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test_cv)
accuracy_score(y_test, y_pred)

But model.score(...) is a shortcut that:

    Calls .predict() internally

    Compares predictions to y_test

    Returns the result as a decimal (e.g., 0.976 for 97.6% accuracy)

✅ Example Output

0.9825918762088972

Means the model correctly classified ~98.26% of test messages as either spam or ham.

```python
from sklearn.metrics import classification_report

y_pred = model.predict(X_test_cv)

print(classification_report(y_test, y_pred))
```

🧠 Explanation

This block is used for detailed model evaluation beyond just accuracy.
✅ Step-by-step:

    model.predict(X_test_cv)

        Uses the trained model to make predictions on the test data.

    classification_report(y_test, y_pred)

        Compares actual vs. predicted labels.

        Returns precision, recall, and F1-score for each class (spam and ham).

📊 Sample Output (Example)

              precision    recall  f1-score   support

         ham       0.99      1.00      1.00      965
        spam       0.98      0.92      0.95      150

    accuracy                           0.99     1115
   macro avg       0.98      0.96      0.97     1115
weighted avg       0.99      0.99      0.99     1115

🔍 What These Metrics Mean
Metric	Meaning
Precision	Of all predicted spam, how many were actually spam?
Recall	Of all actual spam messages, how many did the model correctly catch?
F1-score	Harmonic mean of precision and recall (overall effectiveness).
Support	Actual number of samples per class in the test set.
✅ Interpretation

If you're building a spam filter:

    High recall on spam is crucial (you want to catch all spam).

    But also balance with precision (don’t misclassify important emails as spam).
