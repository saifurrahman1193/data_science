Here's a **simple Python implementation of KNN from scratch** for your example:

---

### 🧮 **Problem:**

We have:

```python
X = [[1], [2], [4], [6], [7]]
y = [0, 0, 1, 1, 1]
```

We want to classify:

```python
x_test = [3]
```

using **k = 3** and **Euclidean distance**.

---



````python
import numpy as np
from collections import Counter

# Training data
X = [[1], [2], [4], [6], [7]]
y = [0, 0, 1, 1, 1]

# Test point
x_test = [3]
k = 3

# Step 1: Compute Euclidean distances
distances = []
for xi, label in zip(X, y):
    dist = np.linalg.norm(np.array(xi) - np.array(x_test))
    distances.append((dist, label))

# Step 2: Sort distances and get k nearest labels
distances.sort()  # Sort by distance
k_nearest_labels = [label for _, label in distances[:k]]

# Step 3: Majority vote
prediction = Counter(k_nearest_labels).most_common(1)[0][0]

# Show result
print("Predicted class for", x_test, "is:", prediction)
````

## 📊 KNN Visualization (1D)

````python
import matplotlib.pyplot as plt
import numpy as np

# Training data
X = [1, 2, 4, 6, 7]
y = [0, 0, 1, 1, 1]
colors = ['blue' if label == 0 else 'red' for label in y]

# Test point
x_test = 3
k = 3

# Compute distances
distances = [(abs(x - x_test), x, label) for x, label in zip(X, y)]
distances.sort()
k_neighbors = distances[:k]

# Plot all points
plt.figure(figsize=(10, 2))
plt.scatter(X, [0]*len(X), c=colors, s=100, label='Training Points')

# Highlight nearest neighbors
for _, x, label in k_neighbors:
    plt.scatter(x, 0, edgecolor='black', facecolor='green', s=300, linewidth=2, label='Neighbor' if x == k_neighbors[0][1] else "")

# Plot test point
plt.scatter(x_test, 0, c='orange', s=200, marker='X', label='Test Point (x=3)')

# Annotate
for i, (x, label) in enumerate(zip(X, y)):
    plt.text(x, 0.1, f'{label}', ha='center', fontsize=12)

plt.yticks([])
plt.xlabel("Feature value (X)")
plt.title("KNN Classification: Predict class for x = 3")
plt.legend()
plt.grid(True, axis='x', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
````

Great! Here's how to **visualize the KNN process in a graph** for your example using `matplotlib`.

We'll plot:

* The training points
* The test point (`x = 3`)
* Highlight the `k = 3` nearest neighbors

---

### 🖼️ What This Graph Shows:

* 🔵 Blue = Class 0
* 🔴 Red = Class 1
* 🟧 Orange = Test point (x = 3)
* 🟩 Green border = The **k = 3 nearest neighbors**

This visual helps you **see the vote decision** — and understand **why the prediction is 0**.
