```python
iloc

    Integer-based indexing: Uses integer positions to select rows and columns.
    Zero-based: Starts counting from 0.
    Slicing: Can use slicing operations with integers.
    Syntax: df.iloc[row_indexer, column_indexer]

loc

    Label-based indexing: Uses column and index names to select data.
    Inclusive: Includes both the start and end points in slicing.
    Slicing: Can use slicing operations with labels.
    Syntax: df.loc[row_indexer, column_indexer]
```

```python
# .iloc: This is a Pandas method used for integer-based indexing.
# [0]: Selects the row at index 0 (which is the first row).
reviews.iloc[0]
```

```python
# .iloc: This is a Pandas method used for integer-based indexing.
# [:, 0]: This specifies the slicing of the DataFrame:

#     : selects all rows.
#     0 selects the first column.
reviews.iloc[:, 0]
```

```python
# [:3, 0]: Specifies the slicing of the DataFrame:

#     :3: Selects rows from the beginning up to, but not including, row 3 (i.e., rows 0, 1, and 2).
#     0: Selects the first column.
reviews.iloc[:3, 0]
```

```python
# [1:3, 0]: Specifies the slicing of the DataFrame:

#     1:3: Selects rows from index 1 (inclusive) up to, but not including, index 3 (i.e., rows 1 and 2).
#     0: Selects the first column.
reviews.iloc[1:3, 0]
```

```python
    # [[0, 1, 2], 0]: Specifies the slicing of the DataFrame:
    #     [0, 1, 2]: Selects rows at indices 0, 1, and 2.
    #     0: Selects the first column.

reviews.iloc[[0, 1, 2], 0]
```

```python
# [-5:]: Selects rows starting from the fifth row from the end (inclusive) to the end of the DataFrame.
reviews.iloc[-5:]
```

```python
# .loc: Label-based location indexing method in Pandas.
# [0, 'country']: Specifies the location of the value you want to access:

#     0: Selects the row with the label (or index) 0.
#     'country': Selects the column named 'country'.
reviews.loc[0, 'country']
```

```python
# .loc: Label-based location indexing method in Pandas.
# [:, ['taster_name', 'taster_twitter_handle', 'points']]: Specifies the slicing of the DataFrame:

#     :: Selects all rows.
#     ['taster_name', 'taster_twitter_handle', 'points']: Selects the columns named 'taster_name', 'taster_twitter_handle', and 'points'.
reviews.loc[:, ['taster_name', 'taster_twitter_handle', 'points']]
```

```python
# The "title" column, which was previously a regular column, becomes the row labels.
# sets the "title" column as the index of the DataFrame
reviews.set_index("title")
```

```python
# selects rows from the DataFrame reviews where the value in the 'country' column is 'Italy'
# reviews.country == 'Italy' 
# or reviews['country'] == 'Italy'
reviews.country == 'Italy'
```

```python
# reviews.country: This part accesses the 'country' column of the DataFrame reviews.
# reviews.country == 'Italy': This creates a Boolean Series where each element is True if the corresponding country is 'Italy', 
# otherwise False.
# reviews.loc[ ]: This is using the .loc indexer to select rows based on conditions.
# reviews.loc[reviews.country == 'Italy']: This combines the above steps, using the Boolean Series created in step 2 to select rows 
# where the condition is True

# reviews.loc[reviews.country == 'Italy'] 
# or reviews[reviews['country'] == 'Italy'] 
# or reviews.query("country == 'Italy'") 
# or reviews[reviews['country'].isin(['Italy'])]

reviews.loc[reviews.country == 'Italy']
```

```python
# reviews.country == 'Italy': This part remains the same, filtering for rows where the 'country' column equals 'Italy'.
# reviews.points >= 90: This selects rows where the 'points' column is greater than or equal to 90.
# &: The & symbol represents the logical AND operator. It ensures that only rows meeting both conditions (Italy and high points) are included.
# reviews.loc[...]: This uses .loc to select rows based on the combined Boolean conditions.
reviews.loc[(reviews.country == 'Italy') & (reviews.points >= 90)]
```

```python
# reviews.loc[(reviews.country == 'Italy') | (reviews.points >= 90)]
# or reviews[(reviews['country'] == 'Italy') | (reviews['points'] >= 90)]
reviews.query("country == 'Italy' or points >= 90")
```

```python
# reviews.loc[reviews.country.isin(['Italy', 'France'])]
# or 
# countries = ['Italy', 'France']
# reviews[reviews['country'].isin(countries)]
# or reviews.query("country in @countries")
# or reviews[reviews['country'].astype(str).isin(countries)]
reviews.loc[reviews.country.isin(['Italy', 'France'])]
```
