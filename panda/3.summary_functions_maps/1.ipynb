```python
The output of reviews.points.describe() typically includes the following statistics:

    count: Number of non-null values in the points column.
    mean: Average of the points values.
    std: Standard deviation of the points values.
    min: Minimum value in the points column.
    25%: First quartile (25th percentile) of the points values.
    50%: Median (50th percentile) of the points values.
    75%: Third quartile (75th percentile) of the points values.
    max: Maximum value in the points column.
```

```python
# reviews.points.map(lambda p: p - review_points_mean) applies a function (lambda function) element-wise to each value in the points column.

#     The lambda function takes a single argument p which represents each point value.
#     It subtracts the previously calculated review_points_mean from each point value p
review_points_mean = reviews.points.mean()
reviews.points.map(lambda p: p - review_points_mean)
```
