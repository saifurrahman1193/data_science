```python
import pandas as pd
reviews = pd.read_excel("./reviews.xlsx")
pd.set_option('display.max_rows', 5)
```

```python
reviews.groupby('points'): This part groups the DataFrame reviews based on the values in the points column. 
In essence, it creates subsets of the data where each subset contains rows with the same points value.

.points: This specifies that we're interested in the points column within each of the created groups.

.count(): This counts the number of rows (or occurrences) within each of the points groups.
```

```python
# Counts how many times each unique point value appears in the dataset.
# reviews['points'].value_counts() 
# or reviews.groupby('points').points.count() 
# or reviews.groupby('points').size() 
# or reviews.pivot_table(index='points', aggfunc='size')

reviews.points
reviews.groupby('points').points.count()
```

```python
# reviews.groupby('points')
# : Groups the DataFrame reviews by the points column, creating subsets for each unique point value.
# .price: Selects the price column within each group.
# .min(): Calculates the minimum value (price) for each group.
# or
# min_prices = reviews.groupby('points')['price'].apply(lambda x: x.min())
# print(min_prices)
# or
# min_prices = reviews.groupby('points')['price'].agg('min')
# print(min_prices)
# or
# min_prices = reviews.pivot_table(values='price', index='points', aggfunc='min')
# print(min_prices)
    
reviews.groupby('points').price.min()
```

```python
# reviews.groupby(['country', 'province'])
# : This part groups the DataFrame reviews based on the values in both the country and province columns, creating subsets for each unique combination of country and province.
# .description: This specifies that we're interested in the description column within each group.
# .agg([len]): This applies the len function (which calculates the length of a string) to the description column within each group.

reviews.groupby(['country', 'province']).description.agg([len])
```
