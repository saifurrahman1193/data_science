```python
import pandas as pd
```

```python
# Creating DataFrames
# 1️⃣ From a Dictionary

data = {'Name': ['Alice', 'Bob', 'Charlie'], 'Age': [25, 30, 35], 'City': ['NY', 'LA', 'SF']}
df = pd.DataFrame(data)
print(df)
```

```python
# From a CSV File
df = pd.read_csv("./data.csv")  # Load CSV into a DataFrame
```

```python
# 3️⃣ From an Excel File

df = pd.read_excel('data.xlsx')
```

```python
# 🔍 Exploring Data

df.head()  # First 5 rows
df.tail()  # Last 5 rows
df.info()  # Summary of DataFrame
df.describe()  # Summary statistics
df.shape  # (rows, columns)
df.columns  # List of column names
df.dtypes  # Data types of each column
```

```python
# 🎯 Selecting Data
# 1️⃣ Selecting Columns

df['Name']  # Single column (Series)
df[['Name', 'Age']]  # Multiple columns (DataFrame)
```

```python
# 2️⃣ Selecting Rows (Indexing)

df.iloc[0]  # First row (by position)
df.loc[0]  # First row (by label)
df.loc[1:2, ['Name', 'Age']]  # Specific rows & columns
```

```python
# 3️⃣ Filtering Data

df[df['Age'] > 25]  # Filter rows where Age > 25
df[df['City'] == 'NY']  # Filter rows where City = 'NY'
```

```python
# 🛠️ Modifying Data
# 1️⃣ Adding a Column

df['Salary'] = [50000, 60000, 70000]
```

```python

# 2️⃣ Updating Values

df.loc[0, 'Age'] = 26  # Update a specific value
df['Age'] = df['Age'] + 1  # Increment all ages by 1
```

```python
# 3️⃣ Removing Columns

df.drop(columns=['Salary'], inplace=True)
```

```python
# 4️⃣ Removing Rows

df.drop(index=1, inplace=True)  # Remove row with index 1
```

```python
# 📌 Sorting & Grouping
# 1️⃣ Sorting Data
df.sort_values(by='Age', ascending=False)
```

```python
# 2️⃣ Grouping Data

df.groupby('City')['Age'].mean()  # Average age per city
```

```python
# 📈 Working with Missing Data

df.isnull().sum()  # Check missing values
df.fillna(0, inplace=True)  # Replace missing values with 0
df.dropna(inplace=True)  # Remove rows with missing values
```

```python
# 📤 Exporting Data

df.to_csv('output.csv', index=False)
df.to_excel('output.xlsx', index=False)
```
