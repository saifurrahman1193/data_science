```python
import numpy as np

# Define vectors
v1 = np.array([2, 3]) # 1-dimensional NumPy array with elements [2, 3]
v2 = np.array([4, 1])

# Vector addition
v_add = v1 + v2
print("Addition:", v_add)

# Magnitude
magnitude = np.linalg.norm(v1)
print("Magnitude of v1:", magnitude)

# Unit vector
unit_vector = v1 / magnitude
print("Unit Vector of v1:", unit_vector)
```
